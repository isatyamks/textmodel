```python
from reportlab.lib.pagesizes import letter
from reportlab.pdfgen import canvas

```

```python
clean_file = "..\\data\\clean\\raw_01.txt"
filename = "..\\data\\pdf\\summary.pdf"
```

```python
def generate_pdf(summary_text, filename):
    c = canvas.Canvas(filename, pagesize=letter)
    c.setFont("Helvetica", 12)
    c.setFont("Helvetica-Bold", 16)
    c.drawString(100, 750, "Lecture Summary")
    c.setFont("Helvetica", 12)
    y_position = 720
    for line in summary_text.split(". "):
        c.drawString(100, y_position, line.strip() + ".")
        y_position -= 20
        
    c.save()
    print(f"PDF generated: {filename}")

```

```python
def read_txt_file(raw_file):
    with open(raw_file, 'r', encoding='utf-8') as file:
        raw_text = file.read()
    return raw_text
```

```python
summary_text = read_txt_file(clean_file)
```

```python
generate_pdf(summary_text,filename)
```
