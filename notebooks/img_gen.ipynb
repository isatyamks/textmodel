```python
import os
import random
from PIL import Image, ImageDraw, ImageFont
import textwrap
```

```python
summary_file = "..\\api\\test_data.txt"
folder_path = "..\\backgrounds"
```

```python
def read_txt_file(raw_file):
    with open(raw_file, 'r', encoding='utf-8') as file:
        raw_text = file.read()
    return raw_text

summary_text = read_txt_file(summary_file)

```

```python

def generate_instagram_post(
    title, subheadings, body, 
    bg_color=None, bg_image_path=None, 
    text_color="white", font_path="C:/Windows/Fonts/arial.ttf",
    output_file="instagram_post.png"
):
    """
    Generate a beautiful Instagram post with custom title, subheadings, and body.

    Parameters:
    - title: Main heading
    - subheadings: List of subheadings
    - body: Main text content
    - bg_color: Background color (e.g., "#3498db" for blue)
    - bg_image_path: Path to a background image (if provided, this will be used instead of bg_color)
    - text_color: Color of text
    - font_path: Path to a .ttf font file
    - output_file: Filename for the generated image
    """

    # Set image dimensions (Instagram post size: 1080x1080)
    width, height = 1080, 1080

    # Load background (Solid color or Image)
    if bg_image_path and os.path.exists(bg_image_path):
        img = Image.open(bg_image_path).resize((width, height))
    else:
        img = Image.new("RGB", (width, height), color=bg_color if bg_color else "#000000")

    draw = ImageDraw.Draw(img)

    # Load fonts
    try:
        title_font = ImageFont.truetype(font_path, 60)
        sub_font = ImageFont.truetype(font_path, 45)
        body_font = ImageFont.truetype(font_path, 35)
    except OSError:
        title_font = ImageFont.load_default()
        sub_font = ImageFont.load_default()
        body_font = ImageFont.load_default()

    # Title Positioning
    y_position = 80
    draw.text((80, y_position), title, fill=text_color, font=title_font)
    y_position += 80

    # Subheadings
    for sub in subheadings:
        draw.text((80, y_position), sub, fill=text_color, font=sub_font)
        y_position += 60

    # Body Text Wrapping
    wrapped_text = textwrap.fill(body, width=40)
    y_position += 40  # Space after subheadings
    for line in wrapped_text.split("\n"):
        draw.text((80, y_position), line, fill=text_color, font=body_font)
        y_position += 50

    # Save the image
    img.save(output_file)
    print(f"Instagram post saved as: {output_file}")

# Example usage
title = "The Rise of AI in 2024"
subheadings = ["Machine Learning Revolution", "Impact on Jobs"]
body = "Artificial Intelligence is transforming industries at an unprecedented pace. From automating tasks to making life easier, AI is becoming an integral part of our daily lives. In 2024, we are witnessing the biggest AI revolution ever."

generate_instagram_post(
    title=title,
    subheadings=subheadings,
    body=body,
    bg_color="#1E1E1E",  # Dark theme background
    text_color="white",
    bg_image_path="backgrounds/sample.jpg",  # Use an image or leave None for solid color
    output_file="ai_instagram_post.png"
)
```
