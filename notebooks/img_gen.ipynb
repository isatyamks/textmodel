```python
from PIL import Image, ImageDraw, ImageFont
import textwrap

def generate_instagram_post(summary_text, filename="summary_post.png"):
    # Image size (Instagram post: 1080x1080 pixels)
    img = Image.new("RGB", (1080, 1080), color=(255, 255, 255))  # White background
    draw = ImageDraw.Draw(img)

    # Load a font (Make sure you have a TTF font file)
    font_path = "/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf"  # Change if needed
    font = ImageFont.truetype(font_path, 40)  # Font size 40

    # Title
    title_font = ImageFont.truetype(font_path, 60)
    draw.text((100, 100), "Lecture Summary", fill=(0, 0, 0), font=title_font)

    # Wrap text to fit within the image width
    wrapped_text = textwrap.fill(summary_text, width=40)  # Adjust width for better fit

    # Draw wrapped text
    y_position = 200  # Start position for text
    for line in wrapped_text.split("\n"):
        draw.text((100, y_position), line, fill=(0, 0, 0), font=font)
        y_position += 60  # Adjust line spacing

    # Save image
    img.save(filename)
    print(f"Instagram post image saved as: {filename}")

# Example usage
summary = "This is a summary of the lecture. It contains key points and insights. The summary is concise and helps in quick revision."
generate_instagram_post(summary)
```
