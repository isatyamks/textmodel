```python
import os
import random
from PIL import Image, ImageDraw, ImageFont
import textwrap
```

```python
summary_file = "..\\api\\test_data.txt"
folder_path = "..\\backgrounds"
```

```python
def read_txt_file(raw_file):
    with open(raw_file, 'r', encoding='utf-8') as file:
        raw_text = file.read()
    return raw_text

summary_text = read_txt_file(summary_file)
```

```python

def generate_instagram_post(summary_text, folder_path, filename="summary_post.png"):
    backgrounds = [f for f in os.listdir(folder_path) if f.endswith((".png", ".jpg", ".jpeg"))]
    if not backgrounds:
        print("No background images found! Using a white background.")
        img = Image.new("RGB", (1080, 1080), color=(255, 255, 255)) 
    else:
        bg_image = random.choice(backgrounds)
        img = Image.open(os.path.join(folder_path, bg_image))
        img = img.resize((1080, 1080))

    draw = ImageDraw.Draw(img)
    font_path = "C:/Windows/Fonts/arial.ttf"  
    try:
        font = ImageFont.truetype(font_path, 40)
        title_font = ImageFont.truetype(font_path, 60)
    except OSError:
        font = ImageFont.load_default()
        title_font = font  
    draw.text((100, 100), "Lecture Summary", fill=(255, 255, 255), font=title_font)
    wrapped_text = textwrap.fill(summary_text, width=40)
    y_position = 200  
    for line in wrapped_text.split("\n"):
        draw.text((100, y_position), line, fill=(255, 255, 255), font=font)
        y_position += 60  
    img.save(filename)
```

```python
summary = "This is a summary of the lecture. It contains key points and insights. The summary is concise and helps in quick revision."
generate_instagram_post(summary,folder_path)
```
