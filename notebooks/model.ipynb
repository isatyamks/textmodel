```python
import spacy
import re
import nltk
from nltk.corpus import stopwords
```

```python
# nltk.download("stopwords")
nlp = spacy.load("en_core_web_sm")
```

```python
raw_file = "..\\data\\raw\\raw_01.txt"
clean = "..\\data\\clean"
```

```python

def read_txt_file(raw_file):
    with open(raw_file, 'r', encoding='utf-8') as file:
        raw_text = file.read()
    return raw_text
```

```python
def clean_text(text):
    filler_words = ["uh", "um", "like", "you know"]
    pattern = r'\b(?:' + '|'.join(filler_words) + r')\b'
    text = re.sub(pattern, '', text, flags=re.IGNORECASE)
    # text = re.sub(r'\s+', ' ', text).strip()
    text = re.sub(r'a'-'z' + r'A'-'Z')
    text = re.sub(r"[^a-zA-Z0-9.,'!? ]", "", text)
    text = re.sub(r',\s*,', ',', text) 
    text = re.sub(r'\s*,\s*', ', ', text)  
    doc = nlp(text)
    cleaned_text = " ".join([token.text for token in doc])
    return cleaned_text
```

```python
import os

raw_text = read_txt_file(raw_file)
print(raw_text)
cleaned_text = clean_text(raw_text)

print(cleaned_text)



```

```python
clean_file_name = os.path.basename(raw_file)
clean_file_path = os.path.join(clean, clean_file_name)

os.makedirs(os.path.dirname(clean_file_path), exist_ok=True)

print("Cleaned Text:\n", cleaned_text)
with open(clean_file_path, 'w', encoding='utf-8') as file:
    file.write(cleaned_text)
```
